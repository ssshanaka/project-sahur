## 6. Loading the Model Weights (Lesson 3.1)

The `.safetensors` file sitting on Hugging Face's servers contains 600 million floating-point numbers. We need to:
1. Download the file.
2. Build an empty PyTorch skeleton from `config.json`.
3. Fill the skeleton with the actual weights ("hydration").
4. Transfer everything into the T4 GPU's VRAM.

We use `AutoModelForCausalLM` (not `AutoModel`) because we need the **Language Modeling Head** — the final layer that outputs logits for next-token prediction. We load in FP16 to halve the VRAM footprint from ~2.4 GB to ~1.2 GB.

```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
import warnings
warnings.filterwarnings("ignore")

MODEL_ID = "Qwen/Qwen3-0.6B"

print(f"📥 Downloading and loading {MODEL_ID}...")

model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float16,   # Half precision: 2 bytes per param instead of 4
    device_map="auto"            # accelerate places weights on the GPU automatically
)

vram_allocated = torch.cuda.memory_allocated() / (1024 ** 3)
print(f"✅ Model loaded!")
print(f"GPU VRAM Allocated: {vram_allocated:.2f} GB")
```

```python
# Print the full architecture tree
print(model)
```

## 7. Loading the Tokenizer (Lesson 3.2)

The model only understands integer tensor IDs. The tokenizer translates human text into token IDs using Byte-Pair Encoding (BPE). Qwen3 has a vocabulary of **151,646 unique tokens**.

The tokenizer also handles **ChatML formatting** — wrapping messages with `<|im_start|>` and `<|im_end|>` special tokens that teach the model who is speaking.

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

# Fix padding if not set (required for batch generation)
if tokenizer.pad_token_id is None:
    tokenizer.pad_token_id = tokenizer.eos_token_id

print(f"Tokenizer loaded!")
print(f"Vocabulary Size:    {tokenizer.vocab_size}")
print(f"EOS Token:          {tokenizer.eos_token} (ID: {tokenizer.eos_token_id})")
print(f"Pad Token:          {tokenizer.pad_token} (ID: {tokenizer.pad_token_id})")
```

### Exploring Tokenization

Let's see how the tokenizer splits text into subword tokens. Notice how common words stay intact, while rarer words get split into subword fragments.

```python
test_text = "Yo Sahur, explain attention mechanisms!"

# Encode: text → token IDs
token_ids = tokenizer.encode(test_text)
print(f"Text:      '{test_text}'")
print(f"Token IDs: {token_ids}")
print(f"Num Tokens: {len(token_ids)}")

# Show each token individually
print("\nToken Breakdown:")
for tid in token_ids:
    print(f"  ID {tid:>6} → {repr(tokenizer.decode([tid]))}")

# Decode: token IDs → text
decoded = tokenizer.decode(token_ids)
print(f"\nDecoded:   '{decoded}'")
```

### ChatML Formatting with `apply_chat_template`

In real inference, we don't manually tokenize raw strings. We structure the conversation into roles (`system`, `user`, `assistant`) and let the tokenizer wrap them in ChatML. The `add_generation_prompt=True` flag appends `<|im_start|>assistant\n` to signal the model that it should now generate a response.

```python
messages = [
    {"role": "system", "content": "You are Sahur, a Gen-Z coding assistant."},
    {"role": "user", "content": "how do I reverse a list in python?"}
]

# Convert to ChatML string (not tokenized yet)
chatml_string = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

print("--- Raw ChatML String ---")
print(chatml_string)
```

## 8. First Generation — The Autoregressive Loop (Lesson 3.3)

An LLM generates text **one token at a time**. `model.generate()` automates this loop:
1. Forward pass through 28 Transformer layers → outputs 151,646 logits.
2. Select the winning token ID.
3. Append to the sequence.
4. Check stop conditions (`<|im_end|>` or `max_new_tokens`).
5. Loop back.

```python
# Tokenize the ChatML prompt and move to GPU
inputs = tokenizer(chatml_string, return_tensors="pt").to(model.device)
prompt_length = inputs["input_ids"].shape[1]

print(f"Prompt token count: {prompt_length}")
print(f"input_ids shape:    {inputs['input_ids'].shape}")
print(f"attention_mask shape: {inputs['attention_mask'].shape}")

# Run the autoregressive loop (greedy decoding, no sampling)
with torch.no_grad():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=100,
        do_sample=False   # Greedy: always pick the highest probability token
    )

# Slice off the prompt tokens to get ONLY the generated response
response_ids = output_ids[0][prompt_length:]
response_text = tokenizer.decode(response_ids, skip_special_tokens=True)

print(f"\nGenerated {len(response_ids)} new tokens.")
print(f"\n🤖 Sahur: {response_text}")
```

## 9. Sampling Strategies (Lessons 3.4–3.7)

Greedy decoding (always picking the highest-probability token) produces deterministic but often boring, repetitive text. Real inference uses **sampling** — randomly selecting from the probability distribution.

- **Temperature** (3.5): Controls randomness. Low (0.2) = focused/deterministic. High (1.5) = creative/chaotic.
- **Top-K** (3.6): Only sample from the K highest-probability tokens. Cuts off the long tail.
- **Top-P / Nucleus** (3.7): Only sample from tokens whose cumulative probability reaches P. Adapts dynamically.

```python
def generate_with_params(prompt_text, temperature=0.8, top_k=50, top_p=0.9, max_tokens=100):
    """Generate a response with specific sampling parameters."""
    msgs = [
        {"role": "system", "content": "You are Sahur, a Gen-Z coding assistant."},
        {"role": "user", "content": prompt_text}
    ]
    formatted = tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
    inp = tokenizer(formatted, return_tensors="pt").to(model.device)
    plen = inp["input_ids"].shape[1]

    with torch.no_grad():
        out = model.generate(
            **inp,
            max_new_tokens=max_tokens,
            do_sample=True,
            temperature=temperature,
            top_k=top_k,
            top_p=top_p,
            repetition_penalty=1.15,
            pad_token_id=tokenizer.pad_token_id
        )

    return tokenizer.decode(out[0][plen:], skip_special_tokens=True)
```

### Experiment: Temperature Comparison

Run the same prompt at different temperatures to see how randomness affects the output.

```python
prompt = "write a one-liner joke about Python programming"

print("--- Temperature 0.2 (Very Focused) ---")
print(generate_with_params(prompt, temperature=0.2))

print("\n--- Temperature 0.8 (Balanced) ---")
print(generate_with_params(prompt, temperature=0.8))

print("\n--- Temperature 1.5 (Very Creative) ---")
print(generate_with_params(prompt, temperature=1.5))
```

## 10. Streaming Inference Engine (Lessons 3.8–3.9)

Standard `model.generate()` is a blocking function — you wait seconds for the full output. `TextStreamer` intercepts the autoregressive loop and prints each token the instant it's generated, just like ChatGPT.

We also enable `use_cache=True` for the **KV Cache** (Lesson 3.8), which prevents the model from recomputing attention for all previous tokens on every step.

```python
from transformers import TextStreamer

class SahurInferenceEngine:
    def __init__(self, model, tokenizer):
        self.model = model
        self.tokenizer = tokenizer

        # skip_prompt=True: Don't echo the user's question back
        self.streamer = TextStreamer(tokenizer, skip_prompt=True, skip_special_tokens=True)

        # Golden Sampling Config (from Lessons 3.4-3.7)
        self.gen_config = {
            "max_new_tokens": 200,
            "do_sample": True,
            "temperature": 0.8,
            "top_k": 50,
            "top_p": 0.90,
            "repetition_penalty": 1.15,
            "pad_token_id": tokenizer.pad_token_id,
            "eos_token_id": tokenizer.eos_token_id,
            "use_cache": True  # KV Cache ON (Lesson 3.8)
        }
        print("✅ SahurInferenceEngine ready.")

    def chat(self, user_message):
        messages = [
            {"role": "system", "content": "You are Sahur, a hyper-responsive Gen Z AI assistant."},
            {"role": "user", "content": user_message}
        ]
        prompt = self.tokenizer.apply_chat_template(
            messages, tokenize=False, add_generation_prompt=True
        )
        inputs = self.tokenizer(prompt, return_tensors="pt").to(self.model.device)

        print("\n🤖 Sahur: ", end="")
        with torch.no_grad():
            self.model.generate(**inputs, streamer=self.streamer, **self.gen_config)
        print("\n" + "─"*50)

engine = SahurInferenceEngine(model, tokenizer)
```

# Module 4 — Exploring the Model

We've loaded the model and run inference. But so far, we've been treating Qwen3 as a black box: prompt in → text out.

Now we're going to **open the box**. We will inspect the model's internal representations (hidden states), visualize where it focuses attention, examine the raw decision scores (logits), convert them to probabilities, measure uncertainty with entropy, and build a step-by-step token prediction inspector.

## 11. Hidden States — The Model's Internal Representations (Lesson 4.1)

As a prompt passes through each of Qwen3's 28 Transformer layers, the token representations evolve. The first layer captures surface-level features (grammar, word identity). The deeper layers capture abstract semantic meaning. We can extract these hidden states by passing `output_hidden_states=True`.

```python
# Set model to evaluation mode (disables dropout)
model.eval()

prompt = "Artificial Intelligence will"
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

with torch.no_grad():
    outputs = model(
        **inputs,
        output_hidden_states=True  # This unlocks hidden state extraction!
    )

hidden_states = outputs.hidden_states  # Tuple of tensors, one per layer

print(f"Number of hidden state layers: {len(hidden_states)} (embedding + 28 decoder layers)")
print(f"Shape of each layer: {hidden_states[0].shape}")
print(f"  → [batch_size, sequence_length, hidden_dim]")
print(f"  → [1, {hidden_states[0].shape[1]} tokens, {hidden_states[0].shape[2]} dimensions]")
```

### Comparing Early vs Deep Layers

Let's compare the representations of the word "Intelligence" at Layer 1 (surface) vs Layer 28 (deepest). We use cosine similarity to measure how much the representation changed.

```python
import torch.nn.functional as F

# Find the token index for "Intelligence" in our prompt
token_ids_list = tokenizer.encode(prompt)
print("Tokens:", [tokenizer.decode([t]) for t in token_ids_list])

# Pick the token index for the first meaningful word (index may vary)
token_idx = 1  # Adjust if needed based on the token breakdown above

# Extract the hidden state for that token at Layer 1 and the last layer
layer_1_vec = hidden_states[1][0, token_idx, :]   # Layer 1
layer_last_vec = hidden_states[-1][0, token_idx, :] # Last layer

# Cosine similarity: 1.0 = identical, 0.0 = unrelated
similarity = F.cosine_similarity(
    layer_1_vec.unsqueeze(0).float(),
    layer_last_vec.unsqueeze(0).float()
).item()

print(f"\nCosine similarity between Layer 1 and Layer {len(hidden_states)-1}: {similarity:.4f}")
print("→ Lower similarity = the model significantly transformed the representation.")
```

## 12. Attention Maps — Where the Model Looks (Lesson 4.2)

Attention weights tell us which tokens the model focused on when processing each position. We extract them by passing `output_attentions=True`. The result is a tensor of shape `[batch, heads, seq_len, seq_len]` — a square matrix showing how much each token "attended to" every other token.

```python
with torch.no_grad():
    attn_outputs = model(
        **inputs,
        output_attentions=True
    )

attentions = attn_outputs.attentions  # Tuple of tensors, one per layer

print(f"Number of attention layers: {len(attentions)}")
print(f"Shape of Layer 0 attention: {attentions[0].shape}")
print(f"  → [batch, num_heads, seq_len, seq_len]")
```

### Visualizing Attention (Layer 0, Head 0)

Let's look at the attention pattern from the very first layer. Each row shows how much attention a token paid to every preceding token.

```python
# Extract Layer 0, Head 0 attention matrix
attn_matrix = attentions[0][0, 0].float().cpu()  # [seq_len, seq_len]
tokens = [tokenizer.decode([t]) for t in token_ids_list]

print(f"Attention Map for Layer 0, Head 0:")
print(f"Tokens: {tokens}\n")

# Print a simple text-based heatmap
header = "         " + " ".join(f"{t:>8}" for t in tokens)
print(header)
for i, row_token in enumerate(tokens):
    row_values = " ".join(f"{attn_matrix[i, j]:.4f}" if j <= i else "   ---  " for j in range(len(tokens)))
    print(f"{row_token:>8} {row_values}")

print("\n→ Higher values = stronger attention between token pairs.")
print("→ '---' = causal mask (can't attend to future tokens).")
```

## 13. Logits — The Raw Decision Scores (Lesson 4.3)

After the prompt passes through all 28 layers, the LM Head outputs a vector of **151,646 raw scores** (one for every token in the vocabulary). These are the **logits** — the model's unnormalized opinion about what should come next. They are NOT probabilities yet.

```python
# We already have the outputs from the forward pass
logits = attn_outputs.logits

print(f"Logits tensor shape: {logits.shape}")
print(f"  → [batch=1, seq_len={logits.shape[1]}, vocab_size={logits.shape[2]}]")

# Extract the logits for the LAST token position (the prediction for the next word)
next_token_logits = logits[0, -1, :]  # Shape: [151646]

print(f"\nPrediction logits shape: {next_token_logits.shape}")
print(f"Min logit: {next_token_logits.min().item():.2f}")
print(f"Max logit: {next_token_logits.max().item():.2f}")

# Top 10 raw logit scores
top_values, top_indices = torch.topk(next_token_logits, k=10)
print(f"\nTop 10 Logit Scores (raw, unnormalized):")
for i in range(10):
    tok = tokenizer.decode([top_indices[i].item()])
    print(f"  {repr(tok):<15} logit: {top_values[i].item():.4f}")
```

## 14. Probabilities & Entropy (Lesson 4.4)

Logits are hard to interpret because they have no fixed scale. **Softmax** converts them into a proper probability distribution that sums to 1.0. 

**Entropy** then tells us how *uncertain* the model is. Low entropy = one token dominates (model is confident). High entropy = many tokens are competitive (model is uncertain).

```python
# Convert logits to probabilities using Softmax
probs = torch.softmax(next_token_logits.float(), dim=-1)

print(f"Probability sum: {probs.sum().item():.6f} (should be ≈ 1.0)")

# Top 10 as actual percentages
top_probs, top_prob_indices = torch.topk(probs, k=10)
print(f"\nTop 10 Predictions (Probability):")
for i in range(10):
    tok = tokenizer.decode([top_prob_indices[i].item()])
    print(f"  {repr(tok):<15} {top_probs[i].item()*100:6.2f}%")

# Calculate Entropy
entropy = -torch.sum(probs * torch.log(probs + 1e-12)).item()
print(f"\n🧠 Entropy: {entropy:.3f}")
print("→ Low entropy = confident. High entropy = uncertain.")
```

## 15. Token Prediction Inspector (Lesson 4.5)

Now we combine everything into a single tool that lets us watch the model think **one token at a time**. Instead of using `model.generate()`, we manually run the forward pass, inspect the logits, compute probabilities, measure entropy, select the token, and append it — for every step.

```python
def run_token_inspector(prompt, max_new_tokens=5, top_k=5):
    """Watch the model predict tokens one at a time."""
    print(f"\n{'='*55}")
    print(f"🔬 TOKEN PREDICTION INSPECTOR")
    print(f"Prompt: '{prompt}'")
    print(f"{'='*55}")

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    generated = inputs["input_ids"]

    for step in range(1, max_new_tokens + 1):
        # 1. Forward pass
        with torch.no_grad():
            outputs = model(input_ids=generated)

        # 2. Extract last-position logits
        next_logits = outputs.logits[:, -1, :]

        # 3. Softmax → probabilities
        probs = torch.softmax(next_logits.float(), dim=-1)

        # 4. Entropy
        entropy = -torch.sum(probs * torch.log(probs + 1e-12), dim=-1).item()

        # 5. Top-K candidates
        values, indices = torch.topk(probs, k=top_k)

        print(f"\n--- STEP {step} ---")
        print("Top Predictions:")
        for i in range(top_k):
            tok = tokenizer.decode([indices[0][i].item()])
            print(f"  {i+1}. {repr(tok):<15} {values[0][i].item()*100:6.2f}%")
        print(f"Entropy: {entropy:.3f}")

        # 6. Greedy selection
        next_token = torch.argmax(probs, dim=-1, keepdim=True)
        chosen = tokenizer.decode(next_token[0])
        print(f"✅ Chosen: {repr(chosen)}")

        # 7. Append and show growing sequence
        generated = torch.cat([generated, next_token], dim=-1)
        current = tokenizer.decode(generated[0], skip_special_tokens=True)
        print(f"📝 Sequence: {current}")

        # 8. EOS check
        if next_token.item() == tokenizer.eos_token_id:
            print("🛑 EOS token reached.")
            break

print("✅ TokenInspector loaded!")
```

### 🧪 Experiment: Factual vs Ambiguous Prompts

Watch how the model's confidence differs between a factual prompt (one correct answer → low entropy) and an ambiguous prompt (many valid continuations → high entropy).

```python
# Factual prompt: model should be very confident
run_token_inspector("The capital of Japan is", max_new_tokens=3)
```

```python
# Ambiguous prompt: model should be uncertain
run_token_inspector("I wonder if", max_new_tokens=5)
```

```python
# Creative prompt: entropy will fluctuate
run_token_inspector("Once upon a time, a dragon", max_new_tokens=5)
```
